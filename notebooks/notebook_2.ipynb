## We have a couple of options to automate hyperparameter combination

1. GridSearchCV is a built-in library in the Python, Scikit-learn documentation library that automates hyperparameter tuning by testing every possible combination of values in a specified grid using cross-validation.

   - It works well for small search spaces, but becomes slow as the number of parameters grows.

2. RandomizedSearchCV it randomly samples combinations. Faster. Better than grid search for larger spaces.

3. Optuna modern AutoML library that intelligently searches the parameter space
