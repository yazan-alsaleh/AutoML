### Why not always use fit_transform()?

Because of data leakage.

This is very important in ML.

Imagine:
```
X_train
X_test
```
You should do:
```
scaler.fit(X_train)

X_train = scaler.transform(X_train)

X_test = scaler.transform(X_test)
```
Why?

Because the model should learn only from training data.
